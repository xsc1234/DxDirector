```python
import json
```

```python
with open("./data/input.json", 'r') as f:
    data_0 = json.load(f)

with open("./output/translate_v2.json", 'r') as f:
    data = json.load(f)

data_1 = [data[i:i + 4] for i in range(0, len(data), 4)]
```

```python
data[0], data_0[0], data_1[0]
```

```python
s = data[0]['input']
s = s.split("<|im_end|>")[-2]
s = s.split("：")[-1]
s
```

```python
def split(s):
    s = s.split("<|im_end|>")[-2]
    s = s.split("：")[-1]
    return s
```

```python
answer = []

for i, line in enumerate(data_1):
    answer.append({
        "original_post" : split(line[0]['input']),
        "zhhant" : line[1]['output'],
        "yue": line[0]['output'],
        "en": line[2]['output'],
    })
```

```python
answer
```

```python
with open("./output/translate_reformat_v2.json", "w+" , encoding='utf-8') as f:
    json.dump(answer, f, ensure_ascii=False, indent=4)
```
